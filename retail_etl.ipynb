```python
import pandas as pd

df = pd.read_csv("../data_raw/Superstore.csv")
df.head()
```

```python
df = pd.read_csv("../data_raw/Superstore.csv", encoding="latin1")
df.head()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['quarter'] = df['order_date'].dt.quarter
```

```python
df['profit_margin'] = df['profit'] / df['sales']
```

```python
df.head()
```

```python
dim_product = df[['product_id', 'product_name', 'category', 'sub-category']].drop_duplicates()
dim_product.head()
```

```python
dim_customer = df[['customer_id', 'customer_name', 'segment']].drop_duplicates()
dim_customer.head()
```

```python
dim_date = df[['order_date', 'year', 'month', 'quarter']].drop_duplicates()
dim_date['date_id'] = dim_date.index + 1
dim_date.head()
```

```python
dim_region = df[['region', 'country', 'state', 'city']].drop_duplicates()
dim_region['region_id'] = dim_region.index + 1
dim_region.head()
```

```python
fact_sales = df.merge(
    dim_date,
    on=['order_date', 'year', 'month', 'quarter'],
    how='left'
)

fact_sales = fact_sales.merge(
    dim_region[['region_id', 'region', 'country', 'state', 'city']],
    on=['region', 'country', 'state', 'city'],
    how='left'
)

fact_sales = fact_sales[['order_id', 'sales', 'profit', 'quantity',
                         'date_id', 'product_id', 'customer_id', 'region_id']]

fact_sales.head()
```

```python
fact_sales.to_csv("../data_cleaned/fact_sales.csv", index=False)
dim_product.to_csv("../data_cleaned/dim_product.csv", index=False)
dim_customer.to_csv("../data_cleaned/dim_customer.csv", index=False)
dim_date.to_csv("../data_cleaned/dim_date.csv", index=False)
dim_region.to_csv("../data_cleaned/dim_region.csv", index=False)
```

```python
print("ETL completed successfully. Clean CSV files are in /data_cleaned/")
```

```python

import openpyxl


tables = {
    'fact_sales': fact_sales,
    'dim_product': dim_product,
    'dim_customer': dim_customer,
    'dim_date': dim_date,
    'dim_region': dim_region
}


dict_rows = []

for table_name, table in tables.items():
    for col in table.columns:
        dict_rows.append({
            'table_name': table_name,
            'column_name': col,
            'data_type': str(table[col].dtype),
            'sample_values': ', '.join(map(str, table[col].head(3).values))
        })


data_dictionary = pd.DataFrame(dict_rows)

# Export as Excel
data_dictionary.to_excel("../data_cleaned/data_dictionary.xlsx", index=False)

print("Data dictionary created successfully at /data_cleaned/data_dictionary.xlsx")
```
